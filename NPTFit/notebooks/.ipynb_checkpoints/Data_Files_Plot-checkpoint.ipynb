# Plot Fermi Data, Exposure, Masks and Templates

```python
%matplotlib inline
%load_ext autoreload
%autoreload 2

import numpy as np
import healpy as hp

import matplotlib
import matplotlib.pyplot as plt
import matplotlib as mpl
```

```python
base_dir = '/zfs/nrodd/NPTFit/Fermi_Data/'
counts=np.load(base_dir + 'FermiData_counts.npy')
exposure=np.load(base_dir + 'FermiData_exposure.npy')
pscmask=np.load(base_dir + 'FermiData_pscmask.npy')
bub=np.load(base_dir + 'Template_bub.npy')
dif=np.load(base_dir + 'Template_dif.npy')
gce=np.load(base_dir + 'Template_gce.npy')
iso=np.load(base_dir + 'Template_iso.npy')
psc=np.load(base_dir + 'Template_psc.npy')
```

```python
nonzero = np.where(counts != 0)[0]
zero = np.where(counts == 0)[0]
counts[nonzero] = np.log10(counts[nonzero])
counts[zero] = -1
hp.mollview(counts,title='Data - counts')
plt.show()
```

```python
hp.mollview(np.log10(exposure),title='Exposure')
plt.show()
```

```python
hp.mollview(pscmask,title='Point Source Catalog Mask')
plt.show()
```

```python
hp.mollview(bub,title='Fermi Bubbles Template')
plt.show()
```

```python
hp.mollview(np.log10(dif),title='Diffuse Model Template (p6v11)')
plt.show()
```

```python
hp.mollview(np.log10(gce),title='Galactic Center Excess Template')
plt.show()
```

```python
hp.mollview(np.log10(iso),title='Isotropic Emission Template')
plt.show()
```

```python
hp.mollview(np.log10(psc),title='Point Source Catalog Template')
plt.show()
```
