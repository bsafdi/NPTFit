# Example of Different Masks

```python
%matplotlib inline
%load_ext autoreload
%autoreload 2

import numpy as np
import healpy as hp

import matplotlib
import matplotlib.pyplot as plt
import matplotlib as mpl

import sys, os
sys.path.append("/zfs/nrodd/NPTFit/Full_Testing/Cleanup")

import create_mask as cm

nside = 128
```

# Example 1: Mask Nothing

```python
example1 = cm.make_mask_total(nside)
hp.mollview(example1,title='',cbar=False,min=0,max=1)
```

# Example 2: Band Mask

```python
example2 = cm.make_mask_total(nside, 
                              band_mask = True, band_mask_range = 30)
hp.mollview(example2,title='',cbar=False,min=0,max=1)
```

# Example 3: Mask outside a box in b and l

```python
example3a = cm.make_mask_total(nside, 
                               l_mask = False, l_deg_min = -30, l_deg_max = 30, 
                               b_mask = True, b_deg_min = -30, b_deg_max = 30)
hp.mollview(example3a,title='',cbar=False,min=0,max=1)

example3b = cm.make_mask_total(nside, 
                               l_mask = True, l_deg_min = -30, l_deg_max = 30, 
                               b_mask = False, b_deg_min = -30, b_deg_max = 30)
hp.mollview(example3b,title='',cbar=False,min=0,max=1)

example3 = cm.make_mask_total(nside, 
                              l_mask = True, l_deg_min = -30, l_deg_max = 30, 
                              b_mask = True, b_deg_min = -30, b_deg_max = 30)
hp.mollview(example3,title='',cbar=False,min=0,max=1)
```

# Example 4: Ring and Annulus Mask

```python
example4a = cm.make_mask_total(nside, 
                               mask_ring = True, inner = 0, outer = 30, ring_b = 0, ring_l = 0)
hp.mollview(example4a,title='',cbar=False,min=0,max=1)

example4b = cm.make_mask_total(nside, 
                               mask_ring = True, inner = 30, outer = 180, ring_b = 0, ring_l = 0)
hp.mollview(example4b,title='',cbar=False,min=0,max=1)

example4c = cm.make_mask_total(nside, 
                               mask_ring = True, inner = 30, outer = 90, ring_b = 0, ring_l = 0)
hp.mollview(example4c,title='',cbar=False,min=0,max=1)

example4d = cm.make_mask_total(nside, 
                               mask_ring = True, inner = 0, outer = 30, ring_b = 45, ring_l = 45)
hp.mollview(example4d,title='',cbar=False,min=0,max=1)
```

# Example 5: Custom Mask

```python
random_custom_mask = np.random.choice(np.array([True, False]), hp.nside2npix(nside))
example5 = cm.make_mask_total(nside, 
                              custom_mask = random_custom_mask)
hp.mollview(example5,title='',cbar=False,min=0,max=1)
```

# Example 6: Full Analysis Mask including Custom Point Source Catalog Mask

```python
base_dir = '/zfs/nrodd/NPTFit/Fermi_Data/'
pscmask=np.array(np.load(base_dir + 'FermiData_pscmask.npy'), dtype=bool)
example6 = cm.make_mask_total(nside,
                              band_mask = True, band_mask_range = 2,
                              mask_ring = True, inner = 0, outer = 30,
                              custom_mask = pscmask)
hp.mollview(example6,title='',cbar=False,min=0,max=1)
```
