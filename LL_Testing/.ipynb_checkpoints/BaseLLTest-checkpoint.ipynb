```python
# This is code to test the speed on the NPTF LL function

# First we need to setup some random parameters to be fed into the LL evaluation
# For testing we use 1000 values
```

```python
import numpy as np
# Setup basic arrays - setup so can adjust how good the fit is by multiplying xbg and PS_dist
size = 1000
data = np.array(np.random.poisson(50, size=size), dtype='int32')
xbg_PSF_compressed = data*1.0
theta_PS = np.array([1.0,2.5,2.0,1.5,10.0,20.0])
#theta_PS_rescale = np.array([1.0*(20.0/10.0)**(2.0),2.5,2.0,1.5,10.0,20.0])
theta_PS2 = np.array([1.0*(20.0/10.0)**(-2.0),2.5,2.0,1.5,20.0,10.0])
PS_dist_compressed = data*1.0
# import f_ary and df_rho_div_f_ary
f_ary = np.loadtxt('f_ary-128-0.106-10.dat')
df = f_ary[1] - f_ary[0]
f_ary_edge = f_ary - df/2 + 0.0001
rho_ary = np.loadtxt('rho_ary-128-0.106-10.dat')
rho_ary = rho_ary / np.sum(df*f_ary*rho_ary)
df_rho_div_f_ary = df*rho_ary / f_ary
```

```python
# First compare 2 break and test it gives the right answer
# Good Fit
import pulsars.likelihood_psf_pdep_2break as llpsfb
import LL
print llpsfb.log_nu_k_ary_PSF_exact_1_PS_2break(0.95*xbg_PSF_compressed, theta_PS, f_ary, 
                                                df_rho_div_f_ary, 0.05*PS_dist_compressed, data)
print LL.log_like(0.95*xbg_PSF_compressed, [theta_PS2], f_ary, df_rho_div_f_ary, [0.05*PS_dist_compressed], data)
# Bad Fit - I checked in this case p_k = 0 for 1000 pixels and so we are just adding the negative penalty
print llpsfb.log_nu_k_ary_PSF_exact_1_PS_2break(0.5*xbg_PSF_compressed, theta_PS, f_ary, 
                                                df_rho_div_f_ary, 0.5*PS_dist_compressed, data)
print LL.log_like(0.5*xbg_PSF_compressed, [theta_PS2], f_ary, df_rho_div_f_ary, [0.5*PS_dist_compressed], data)
```

```python
# Test the 1 break case
import LL
import pulsars.likelihood_psf_pdep_new as llpsf
theta_PS = np.array([1.0,2.5,1.5,10.0])
# Good Fit
print llpsf.log_nu_k_ary_PSF_exact_1_PS(0.95*xbg_PSF_compressed, theta_PS, f_ary, df_rho_div_f_ary, 0.05*PS_dist_compressed, data)
print LL.log_like(0.95*xbg_PSF_compressed, [theta_PS], f_ary, df_rho_div_f_ary, [0.05*PS_dist_compressed], data)
# Bad Fit - I checked in this case p_k = 0 for 1000 pixels and so we are just adding the negative penalty
print llpsf.log_nu_k_ary_PSF_exact_1_PS(0.5*xbg_PSF_compressed, theta_PS, f_ary, df_rho_div_f_ary, 0.5*PS_dist_compressed, data)
print LL.log_like(0.5*xbg_PSF_compressed, [theta_PS], f_ary, df_rho_div_f_ary, [0.5*PS_dist_compressed], data)
```

```python
# Test the 3 break case
import pulsars.likelihood_psf_pdep_3break as llpsfc
theta_PS = np.array([1.0,2.5,2.0,2.0,1.5,10.0,20.0,30.0])
theta_PS3 = np.array([(20.0/10.0)**(-2.0)*(30.0/20.0)**(-2.0)*1.0,2.5,2.0,2.0,1.5,30.0,20.0,10.0])
# Good Fit
print llpsfc.log_nu_k_ary_PSF_exact_1_PS_3break(0.95*xbg_PSF_compressed, theta_PS, f_ary, df_rho_div_f_ary, 0.05*PS_dist_compressed, data)
print LL.log_like(0.95*xbg_PSF_compressed, [theta_PS3], f_ary, df_rho_div_f_ary, [0.05*PS_dist_compressed], data)
# Bad Fit - I checked in this case p_k = 0 for 1000 pixels and so we are just adding the negative penalty
print llpsfc.log_nu_k_ary_PSF_exact_1_PS_3break(0.5*xbg_PSF_compressed, theta_PS, f_ary, df_rho_div_f_ary, 0.5*PS_dist_compressed, data)
print LL.log_like(0.5*xbg_PSF_compressed, [theta_PS3], f_ary, df_rho_div_f_ary, [0.5*PS_dist_compressed], data)
```

```python
# Now compare time - using 2 break
# Good Fit
theta_PS = np.array([1.0,2.5,2.0,1.5,10.0,20.0])
theta_PS2 = np.array([1.0*(20.0/10.0)**(-2.0),2.5,2.0,1.5,20.0,10.0])
%timeit -r 100 llpsfb.log_nu_k_ary_PSF_exact_1_PS_2break(0.95*xbg_PSF_compressed, theta_PS, f_ary, df_rho_div_f_ary, 0.05*PS_dist_compressed, data)
%timeit -r 100 LL.log_like(0.95*xbg_PSF_compressed, [theta_PS2], f_ary, df_rho_div_f_ary, [0.05*PS_dist_compressed], data)
# Bad Fit
%timeit -r 100 llpsfb.log_nu_k_ary_PSF_exact_1_PS_2break(0.5*xbg_PSF_compressed, theta_PS, f_ary, df_rho_div_f_ary, 0.5*PS_dist_compressed, data)
%timeit -r 100 LL.log_like(0.5*xbg_PSF_compressed, [theta_PS2], f_ary, df_rho_div_f_ary, [0.5*PS_dist_compressed], data)
```
